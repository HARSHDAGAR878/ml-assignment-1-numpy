```python
import numpy as np

np.random.seed(42)

A = np.random.normal(40, 12, (1000, 5))

print("Shape:", A.shape)
print("Dtype:", A.dtype)
print("Memory:", A.nbytes, "bytes")

A = A.astype(np.float32)

print("New dtype:", A.dtype)
print("New memory:", A.nbytes, "bytes")
```

```python
mean = A.mean(axis=0)
std = A.std(axis=0)

Z = (A - mean) / std

print("Mean of columns:")
print(Z.mean(axis=0))

print("Standard deviation of columns:")
print(Z.std(axis=0))
```

```python
import numpy as np

def minmax(X, axis):
    minimum = X.min(axis=axis, keepdims=True)
    maximum = X.max(axis=axis, keepdims=True)
    return (X - minimum) / (maximum - minimum)

# Define X using a slice of A as done in the later cells
X = A[:200]

result0 = minmax(X, 0)

print("Axis 0 minimum:", result0.min(axis=0))
print("Axis 0 maximum:", result0.max(axis=0))
result1 = minmax(X, 1)

print("Axis 1 minimum:", result1.min(axis=1))
print("Axis 1 maximum:", result1.max(axis=1))
```

```python
np.random.seed(42)

mask = np.random.random(A.shape) < 0.05

A[mask] = np.nan

column_means = np.nanmean(A, axis=0)

for i in range(A.shape[1]):
    missing = np.isnan(A[:, i])
    A[missing, i] = column_means[i]

print(np.isnan(A).sum() == 0)
```

```python
mean = A.mean(axis=0)
std = A.std(axis=0)

outliers = np.abs(A - mean) > 3 * std

median = np.median(A, axis=0)

for i in range(A.shape[1]):
    A[outliers[:, i], i] = median[i]

print("Outliers replaced")
```

```python
X = A[:200]

X1 = X[:, np.newaxis, :]
X2 = X[np.newaxis, :, :]

difference = X1 - X2

distance_matrix = np.sqrt(
    (difference ** 2).sum(axis=2))
print(distance_matrix.shape)
print(np.allclose(np.diag(distance_matrix), 0))
print(np.allclose(distance_matrix, distance_matrix.T))
```

```python
import numpy as np


distance_matrix_copy = distance_matrix.copy()
np.fill_diagonal(distance_matrix_copy, np.inf)

# finding 5 smallest distance indices for each row
nearest_5_indices = np.argpartition(distance_matrix_copy, kth=5, axis=1)[:, :5]

print("Shape:", nearest_5_indices.shape)
print("Sample of first 5 rows nearest neighbors:\n", nearest_5_indices[:5])
```

```python

%%time
import numpy as np
np.random.seed(42)
numbers = np.random.randint(0, 100, size=1_000_000)

counts_numpy = np.bincount(numbers)
print("First 10 counts (NumPy):", counts_numpy[:10])

numbers_list = numbers.tolist()
counts_python = [0] * 100
for num in numbers_list:
    counts_python[num] += 1

print("First 10 counts (Python):", counts_python[:10])
print("Results match:", list(counts_numpy) == counts_python)
```

```python
import numpy as np
import pandas as pd

from numpy.lib.stride_tricks import sliding_window_view

x = np.random.randn(10000)

windows = sliding_window_view(x, 10)

print("Shape:", windows.shape)

numpy_means = windows.mean(axis=1)

pandas_means = pd.Series(x).rolling(10).mean().dropna().to_numpy()

print("Results are same:", np.allclose(numpy_means, pandas_means))
```
